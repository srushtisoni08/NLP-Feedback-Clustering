```python
import pandas as pd
import re
```

```python
df = pd.read_csv(
    "../data/raw/Student_Feedback.csv",
    encoding="latin-1"
)

print("Original shape:", df.shape)
```

```python
display(df[["comments"]].head(10))
```

```python
def clean_text(text):
    # Convert text to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(r"http\S+|www\S+", "", text)

    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z\s]", " ", text)

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
df["cleaned_text"] = df["comments"].apply(clean_text)

display(
    df[["comments", "cleaned_text"]].head(20)
)
```

```python
print("Original rows:", len(df))
print("Unique raw comments:", df["comments"].nunique())
print("Unique cleaned comments:", df["cleaned_text"].nunique())

print(
    "Duplicates after cleaning:",
    df["cleaned_text"].duplicated().sum()
)
```
