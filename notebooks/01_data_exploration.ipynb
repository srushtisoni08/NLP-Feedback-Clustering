```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(
    "../data/raw/Student_Feedback.csv",
    encoding="latin1"
)

print("Dataset loaded successfully!")
```

```python
print("First 5 rows:")
display(df.head())

print("\nDataset shape:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
print("Sentiment distribution:")
print(df["sentiments"].value_counts())
```

```python
df["sentiments"].value_counts().sort_index().plot(
    kind="bar"
)

plt.xlabel("Sentiment Label")
plt.ylabel("Number of Feedbacks")
plt.title("Sentiment Distribution")
plt.tight_layout()
plt.show()
```

```python
df["text_length"] = df["comments"].astype(str).str.len()

df["text_length"].hist()

plt.xlabel("Text Length (characters)")
plt.ylabel("Number of Feedbacks")
plt.title("Feedback Length Distribution")
plt.tight_layout()
plt.show()
```

```python
df["word_count"] = (
    df["comments"]
    .astype(str)
    .str.split()
    .str.len()
)

df["word_count"].hist()

plt.xlabel("Word Count")
plt.ylabel("Number of Feedbacks")
plt.title("Feedback Word Count Distribution")
plt.tight_layout()
plt.show()
```

```python
print("\nText Statistics:")
print(
    df[
        ["text_length", "word_count"]
    ].describe()
)
```

```python
df.to_csv(
    "../data/processed/eda_data.csv",
    index=False
)
```
